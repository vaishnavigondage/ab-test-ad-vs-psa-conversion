```python
import pandas as pd, numpy as np
from scipy import stats
df = pd.read_csv("marketing_AB.csv")
g = df.groupby("test group")["converted"].agg(["sum","count","mean"])
p1, p0 = g.loc["ad","mean"], g.loc["psa","mean"]
n1, n0 = g.loc["ad","count"], g.loc["psa","count"]
se = np.sqrt(p1*(1-p1)/n1 + p0*(1-p0)/n0)
print(p1, p0, p1-p0, (p1-p0-1.96*se, p1-p0+1.96*se))
pp = (g["sum"].sum())/(n1+n0)
z = (p1-p0)/np.sqrt(pp*(1-pp)*(1/n1+1/n0)); print(z, 2*stats.norm.sf(abs(z)))
```
